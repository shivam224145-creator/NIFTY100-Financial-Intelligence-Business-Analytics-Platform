```python
import pandas as pd
import numpy as np
ml_df = pd.read_csv(
    r"C:\Users\LENOVO\Desktop\nifty100_project\data\clean\company_health_scores.csv"
)

ml_df.head()
```

```python
ml_df.columns
```

```python
# Add Missing ML Score Columns
profit = pd.read_csv(
    r"C:\Users\LENOVO\Desktop\nifty100_project\data\clean\profit_clean.csv"
)

latest_year = ml_df['fiscal_year'].max()

latest_profit = profit[profit['fiscal_year'] == latest_year][[
    'symbol',
    'dividend_payout'
]].drop_duplicates(subset=['symbol'])

ml_df = ml_df.merge(
    latest_profit,
    on='symbol',
    how='left'
)

ml_df['overall_score'] = ml_df['health_score']

ml_df['leverage_score'] = ml_df['debt_score']

ml_df['dividend_score'] = np.where(
    ml_df['dividend_payout'].fillna(0) >= 30, 20,
    np.where(ml_df['dividend_payout'].fillna(0) >= 10, 12, 6)
)

ml_df['growth_score'] = np.where(
    ml_df['net_profit_margin_pct'] >= 20, 20,
    np.where(ml_df['net_profit_margin_pct'] >= 10, 14, 8)
)

ml_df['trend_score'] = np.where(
    ml_df['free_cash_flow'] > 0, 20, 8
)

ml_df.head()
```

```python
ml_df.columns
```

```python
# Final Save
ml_df.to_csv(
    r"C:\Users\LENOVO\Desktop\nifty100_project\data\clean\company_updated_health_scores.csv",
    index=False
)

print("company_updated_health_scores.csv created successfully.")
```
